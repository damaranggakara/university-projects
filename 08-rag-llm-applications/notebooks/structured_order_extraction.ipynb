```python
#Library
from google import genai
```

```python
#API key dan client
client = genai.Client(api_key="INSERT-YOUR-GEMINI-API-KEY-HERE")
```

```python
#Function
def get_menu(menu_type: str) -> dict:
    if menu_type == "food":
        return {
            "nasi goreng": 15000,
            "mie goreng": 12000,
            "sate ayam": 20000
        }
    elif menu_type == "drink":
        return {
            "es teh": 5000,
            "es jeruk": 7000,
            "kopi": 10000
        }
    else:
        return {"error": "Menu type not found"}
```

```python
#Tool
tools = [{
    "function_declarations": [{
        "name": "get_menu",
        "description": "Mengambil daftar menu makanan atau minuman",
        "parameters": {
            "type": "object",
            "properties": {
                "menu_type": {
                    "type": "string",
                    "enum": ["food", "drink"]
                }
            },
            "required": ["menu_type"]
        }
    }]
}]
```

```python
#Input
user_input = "Saya pesan Sate ayam, Nasi goreng, dan es teh. Berapa totalnya?"
```

```python
#System insturction and configuration (tools)
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=user_input,
    config={
        "tools": tools,
        "system_instruction": (
            "Kamu adalah pelayan restoran.\n"
            "WAJIB menggunakan function get_menu.\n"
            "Jika pesanan mengandung makanan dan minuman, ambil KEDUANYA.\n"
            "Jangan menjawab sebelum semua data lengkap.\n"
        )
    }
)
```

```python
#Data function
menu_data = {}
menu_data.update(get_menu("food"))
menu_data.update(get_menu("drink"))
```

```python
#Respond
final_response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=f"""
Pesanan user: {user_input}
Data menu lengkap:
{menu_data}
Tugas:
1. Kenali semua item pesanan (tidak peduli kapital).
2. Ambil harga dari data menu.
3. Hitung total dengan tepat.
4. Tampilkan rincian dan total harga.
Jawab sebagai pelayan restoran yang ramah.
"""
)
```
