```python
#TASK 1
#Isi artikel
article = """
Dampak Ganda E-commerce di Indonesia: Antara Pertumbuhan Ekonomi dan Kesenjangan Digital
Pendahuluan
Ekonomi digital Indonesia, didominasi oleh sektor e-commerce, telah menjelma menjadi lokomotif utama pertumbuhan nasional. Studi e-Conomy SEA memproyeksikan nilai ekonomi digital Indonesia akan mencapai USD 77 miliar pada tahun 2022 dan terus melesat hingga mendekati USD 130 miliar pada tahun 2025. Perdagangan elektronik atau e-commerce menyumbang porsi terbesar dari angka fantastis ini. Namun, di balik angka pertumbuhan yang memukau, muncul tantangan serius terkait pemerataan dan kesiapan sumber daya manusia (SDM) yang menciptakan dampak ganda pada perekonomian.
Pendorong Pertumbuhan dan Akses Pasar
Kehadiran e-commerce memberikan dampak positif yang fundamental, terutama bagi sektor Usaha Mikro, Kecil, dan Menengah (UMKM). Platform seperti marketplace telah menghapus batasan geografis, memungkinkan produk dari pengusaha di daerah terpencil sekalipun untuk diakses oleh konsumen nasional hingga global.
Meningkatkan Konsumsi Domestik: Kemudahan transaksi, promosi yang intensif, dan variasi produk yang luas melalui e-commerce secara langsung meningkatkan daya beli dan konsumsi domestik, yang merupakan komponen penting dalam Produk Domestik Bruto (PDB).
Penciptaan Lapangan Kerja Baru: Pertumbuhan ekosistem e-commerce secara masif menciptakan kebutuhan akan profesi baru, mulai dari kurir logistik, content creator, analis data, hingga prompt engineer—pekerjaan yang tidak ada di era perdagangan konvensional.
Efisiensi Biaya Operasional: Bagi pelaku usaha, biaya pemasaran dan distribusi menjadi lebih efisien. Mereka tidak lagi memerlukan toko fisik di lokasi mahal, melainkan cukup fokus pada kualitas produk dan strategi digital.
Tantangan dan Ancaman Kesenjangan
Meskipun membawa berkah, fenomena e-commerce juga mempertajam masalah struktural dalam perekonomian Indonesia, khususnya terkait kesiapan digital.
Kesenjangan Akses Infrastruktur: Pertumbuhan e-commerce sangat timpang antara wilayah perkotaan (Jawa-Bali) dan luar Jawa. Infrastruktur internet yang belum merata dan biaya logistik yang tinggi di daerah terpencil menjadi penghalang utama bagi UMKM di sana untuk ikut bersaing.
Krisis Talenta Digital: Menurut data Kementerian Kominfo, Indonesia membutuhkan sekitar 600.000 talenta digital baru setiap tahunnya. Kenyataan di lapangan menunjukkan bahwa lulusan yang memiliki keterampilan hard skill (seperti analisis data, keamanan siber, atau prompt engineering) masih jauh dari jumlah yang dibutuhkan, menghambat laju inovasi.
Isu Keamanan dan Regulasi: Peningkatan transaksi juga diikuti dengan peningkatan risiko penipuan daring (cybercrime) dan isu perlindungan data pribadi konsumen. Pemerintah dan regulator dituntut untuk sigap dalam menyusun kerangka hukum yang memadai tanpa mematikan inovasi.
Pergeseran Perilaku Konsumtif: Kemudahan belanja 24/7 dan flash sale memicu perilaku konsumtif yang berlebihan (impulsive buying) di kalangan masyarakat, yang berpotensi mengganggu stabilitas keuangan rumah tangga.
Kesimpulan
E-commerce di Indonesia berada di persimpangan jalan. Ia adalah mesin pertumbuhan yang tak terbantahkan dan telah merevolusi cara bisnis beroperasi. Namun, keberlanjutan dan dampaknya yang inklusif sangat bergantung pada kemampuan negara dalam mengatasi kesenjangan digital, mempercepat pemerataan infrastruktur, dan mengisi defisit kebutuhan talenta digital melalui reformasi pendidikan dan pelatihan. Tanpa intervensi yang tepat, e-commerce hanya akan memperkaya wilayah yang sudah maju dan meninggalkan mereka yang kurang siap.
"""
```

```python
#Prompt
prompt = f"""
Please summarize the following article with these requirements:
1. The summary must use bullet points.
2. The summary must not exceed 500 words.
3. The summary must be written in formal English.
4. Use concise and academic language.
5. Only include information from the article.
Article:
{article}
"""
```

```python
#Generate response dari Gemini
response_summary = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)
#Tampilkan hasil rangkuman
print(response_summary.text)
```

```python
from pathlib import Path
#TASK 2
#Path CSV
csv_path = str(Path.cwd().parent / "data" / "customer_feedback.csv")
#Membaca file CSV
df = pd.read_csv(csv_path)
df.head()
```

```python
#Menggabungkan semua customer feedback menjadi satu text
all_feedback = "\n".join(df["Teks Customer Feedback"].astype(str))
```

```python
#Prompt
prompt_feedback = f"""
You are an expert product analyst specializing in customer feedback analysis.

Analyze the customer feedback data for a newly released perfume product.

Your tasks:
1. Perform sentiment analysis on the customer feedback.
2. Identify the main strengths of the product.
3. Identify the main weaknesses of the product.
4. Provide at least 55 actionable recommendations to improve the product.
5. Use formal English.
6. Organize the response with clear headings and bullet points.
7. Make the analysis detailed and professional.

Customer Feedback Data:
{all_feedback}
"""
```

```python
#Generate response dari Gemini
response_feedback = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt_feedback
)
#Tampilkan hasil rangkuman
print(response_feedback.text)
```
